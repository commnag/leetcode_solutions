```python
# 9. Palindrome Number
def isPalindrome(x) -> bool:
        return str(x) == ''.join(list(str(x))[::-1])

isPalindrome('aba')
```

```python
# 13. Roman to Integer
def romanToInt(s) -> int:
    num = 0
    my_dict = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000, 'IV': 3, 'IX': 8, 'XL': 30, 'XC': 80, 'CD': 300, 'CM': 800}
    for i in range(len(list(s))):
        if s[i] == 'I':
            num += my_dict['I']
        elif s[i] == 'V':
            if i != 0 and s[i-1] == 'I':
                num += my_dict['IV']
            else:
                num += my_dict['V']
        elif s[i] == 'X':
            if i != 0 and s[i-1] == 'I':
                num += my_dict['IX']
            else:
                num += my_dict['X']
        elif s[i] == 'L':
            if i != 0 and s[i-1] == 'X':
                num += my_dict['XL']
            else:
                num += my_dict['L']
        elif s[i] == 'C':
            if i != 0 and s[i-1] == 'X':
                num += my_dict['XC']
            else:
                num += my_dict['C']
        elif s[i] == 'D':
            if i != 0 and s[i-1] == 'C':
                num += my_dict['CD']
            else:
                num += my_dict['D']
        elif s[i] == 'M':
            if i != 0 and s[i-1] == 'C':
                num += my_dict['CM']
            else:
                num += my_dict['M']

    return num

romanToInt('MCMXCIV')
```
