```python
# Longest Commong Prefix

def longestCommonPrefix(ls):
    final_sub = ''
    for index, ch in enumerate(ls[0]):
        for word in ls[1:]:
            if index >= len(word) or word[index] != ch:
                return final_sub
        final_sub += ch 
    return final_sub
longestCommonPrefix(['flower', 'flow', 'flight'])

# Optimal approach
# def longestCommonPrefix(self, strs: list[str]) -> str:
#     first, last = min(strs), max(strs)
#     for i, ch in enumerate(first):
#         if ch != last[i]:
#             return first[:i]
#     return first
```

```python
# Valid Parenthesis

def isValid(s):
    parenthesis = list(s)
    memory = []
    allowable_values = {'(': ['(', '[', '{', ')'],
                        '{': ['(', '[', '{', '}'],
                        '[': ['(', '[', '{', ']'],
                        ')' : [],
                        '}' : [],
                        ']' : []
                        }
    complementary = {
        '(' : ')',
        '[' : ']',
        '{' : '}'
    }
    if len(parenthesis) < 2:
        return False 
    for i in parenthesis: 
        if len(memory) == 0:
            memory.append(i)
        else:
            if i in allowable_values[memory[-1]]:
                if i == complementary[memory[-1]]:
                    memory.pop()
                else:    
                    memory.append(i)
            else:
                return False
    if len(memory) == 0: 
        return True
    else: 
        return False

isValid('((')
```
