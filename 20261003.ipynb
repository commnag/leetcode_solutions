```python
# 26. Remove Duplicates from Sorted Array

def removeDuplicates(nums):
    # [0,0,1,1,1,2,2,3,3,4]
    i = 0
    for j in range(1, len(nums)):
        if nums[i] != nums[j]:
            i += 1
            nums[i] = nums[j]

    return i+1

removeDuplicates([0,0,1,1,1,2,2,3,3,4])


```
