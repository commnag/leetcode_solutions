```python
# 66. Plus One

def plusOne(digits):
    return [int(y) for y in list(str(int(''.join([str(x) for x in digits])) + 1))]

plusOne([9])
```

```python
# 67. Add Binary

def addBinary(a, b):
    a_index, b_index, carry = len(a) - 1, len(b) - 1, 0
    total = []
    while a_index >= 0 and b_index >= 0:
        if (int(a[a_index]) == 0 and int(b[b_index]) == 1) or (int(a[a_index]) == 1 and int(b[b_index]) == 0):
            total.append(1 + carry)
            carry = 0
            a_index -= 1
            b_index -= 1
        elif int(a[a_index]) == 0 and int(b[b_index]) == 0:
            total.append(0 + carry)
            carry = 0
            a_index -= 1
            b_index -= 1
        elif int(a[a_index]) == 1 and int(b[b_index]) == 1:
            total.append(0 + carry)
            carry = 1
            a_index -= 1
            b_index -= 1
    if carry == 1:
        total.append(1)

    return total[::-1]
#[0, 0, 1, 0, 1]

addBinary('11', '1')
```
