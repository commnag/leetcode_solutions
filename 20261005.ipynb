```python
# 66. Plus One

def plusOne(digits):
    return [int(y) for y in list(str(int(''.join([str(x) for x in digits])) + 1))]

plusOne([9])
```

```python
# 67. Add Binary

def addBinary(a, b):

    a_index, b_index, carry = len(a) - 1, len(b) - 1, 0
    result = []

    while a_index >= 0 or b_index >= 0 or carry:
        total = carry 
        if a_index >= 0:
            total += int(a[a_index])
            a_index -= 1
        if b_index >= 0:
            total += int(b[b_index])
            b_index -= 1
        result.append(str(total % 2))
        carry = total // 2

    return ''.join(reversed(result))


```
