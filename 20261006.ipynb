```python
# 69. Sqrt(x)

def mySqrt(x):
    lower = 0
    upper = x
    mid = None

    while (upper - lower) > 1:
        mid = (upper + lower) // 2
        if (mid * mid) <= x:
            lower = mid
        else: 
            upper = mid
          
    if x < upper * upper:
        return lower 
    else:
        return upper

mySqrt(168)
```
